```python
# -----------------------------------
# Alliance Classification (String)
# -----------------------------------

# Official NATO member countries (can be extended)
nato_countries = [
    "United States",
    "United Kingdom",
    "Germany",
    "France",
    "Canada",
    "Italy",
    "Spain",
    "Turkey",
    "Poland",
    "Netherlands",
    "Belgium",
    "Norway",
    "Denmark",
    "Portugal",
    "Greece"
]

# Optional neutral / special countries
others_countries = [
    "Switzerland",
    "Austria"
]

def assign_alliance(country):
    if country in nato_countries:
        return "NATO"
    elif country in others_countries:
        return "Others"
    else:
        return "Non-NATO"

df["alliance"] = df["country"].apply(assign_alliance)
df["alliance"] = df["alliance"].astype("string")
```
