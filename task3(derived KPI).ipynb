```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv(r"C:\Users\avach\Downloads\processed_dataset.csv")
work_df = df.copy()
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
    .str.replace("__", "_")
)
df.columns
```

```python
df[["total_aircraft", "tanks", "naval_assets", "population"]].dtypes
```

```python
df["total_aircraft"].head()
df["total_aircraft"].dtype
```

```python
numeric_cols = [
    "total_aircraft", "tanks", "naval_assets",
    "population", "defense_budget_usd", "gdp_usd",
    "total_personnel", "active_personnel",
    "land_area_sq_km", "coastline_km", "power_index_rank"
]

for col in numeric_cols:
    if col in work_df.columns:
        work_df[col] = (
            work_df[col]
            .astype(str)
            .str.replace(",", "", regex=False)
            .str.replace("+", "", regex=False)
            .str.strip()
        )
        work_df[col] = pd.to_numeric(work_df[col], errors="coerce")
```

```python
df.head()
```

```python
#Assets per Capita
work_df["assets_per_capita"] = (
    work_df["total_aircraft"] +
    work_df["tanks"] +
    work_df["naval_assets"]
) / work_df["population"]
```

```python
work_df[["country", "assets_per_capita"]].head()
```

```python
#Defense Budget to GDP Ratio
work_df["defense_budget_to_gdp"] = (
    work_df["defense_budget_usd"] / work_df["gdp_usd"]
)
```

```python
work_df[["country", "assets_per_capita","defense_budget_to_gdp"]].head()
```

```python
#Personnel Density
work_df["personnel_density"] = (
    work_df["total_personnel"] / work_df["population"]
)
```

```python
work_df[["country", "personnel_density"]].head()
```

```python
#Defense Budget per Soldier
work_df["defense_budget_per_soldier"] = (
    work_df["defense_budget_usd"] / work_df["active_personnel"]
)
```

```python
work_df[["country", "defense_budget_per_soldier"]].head()
```

```python
#Power Index Rank Gap (USA reference)
usa_rank = work_df.loc[
    work_df["country"] == "United States",
    "power_index_rank"
].iloc[0]

work_df["power_index_rank_gap"] = (
    work_df["power_index_rank"] - usa_rank
)
```

```python
work_df[["country", "power_index_rank_gap"]].head()
```

```python
#Air Power Ratio
work_df["air_power_ratio"] = (
    work_df["total_aircraft"] / work_df["active_personnel"]
)
```

```python
work_df[["country", "air_power_ratio"]].head()
```

```python
#Armor Intensity Index
work_df["armor_intensity_index"] = (
    work_df["tanks"] / work_df["land_area_sq_km"]
)
```

```python
work_df[["country", "armor_intensity_index"]].head()
```

```python
#Naval Strength per Coastline
work_df["naval_strength_per_coastline"] = (
    work_df["naval_assets"] / work_df["coastline_km"]
)
```

```python
work_df[["country", "naval_strength_per_coastline"]].head()
```

```python
#Military Burden Index
work_df["military_burden_index"] = (
    work_df["defense_budget_usd"] / work_df["population"]
)
```

```python
work_df[["country", "military_burden_index"]].head()
```

```python
#Coalition Strength Index (dynamic)
work_df["total_assets"] = (
    work_df["total_aircraft"] +
    work_df["tanks"] +
    work_df["naval_assets"]
)
```

```python
work_df[["country", "total_assets"]].head()
```

```python
coalition_strength = work_df.loc[
    work_df["alliance"] == "NATO",
    "total_assets"
].sum() 
```

```python
KPI_df = work_df[
    [
        "country",
        "year",
        "assets_per_capita",
        "defense_budget_to_gdp",
        "personnel_density",
        "defense_budget_per_soldier",
        "power_index_rank_gap",
        "air_power_ratio",
        "armor_intensity_index",
        "naval_strength_per_coastline",
        "military_burden_index",
        "total_assets"
    ]
]

KPI_df.to_csv("military_derived_metrics(KPI).csv", index=False)
```
