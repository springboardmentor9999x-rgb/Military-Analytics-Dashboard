```python
import pandas as pd
import numpy as np

# Load data
df = pd.read_csv('military_cleaned.csv')

# --- KPI CALCULATION SECTION ---

df['assets_per_capita'] = (
    df['total_aircraft'] + df['tanks'] + df['naval_assets']
) / df['population']

df['budget_to_gdp_ratio'] = df['defense_budget_usd'] / df['gdp_usd']

df['personnel_density'] = df['total_personnel'] / df['population']

df['budget_per_soldier'] = df['defense_budget_usd'] / np.where(
    df['active_personnel'] == 0, 1, df['active_personnel']
)

df['power_index_rank_gap'] = df['power_index_rank'] - 1

df['air_power_ratio'] = df['total_aircraft'] / np.where(
    df['active_personnel'] == 0, 1, df['active_personnel']
)

df['armor_intensity_index'] = df['tanks'] / df['land_area_sq_km']

df['naval_strength_per_coastline'] = df['naval_assets'] / np.where(
    df['coastline_km'] == 0, 1, df['coastline_km']
)

df['military_burden_index'] = df['defense_budget_usd'] / df['population']

df['total_assets'] = (
    df['total_aircraft'] + df['tanks'] + df['naval_assets']
)

# --- CLEANING SECTION ---

df = df.replace([np.inf, -np.inf], 0).fillna(0)

ratio_cols = [
    'assets_per_capita',
    'budget_to_gdp_ratio',
    'personnel_density',
    'budget_per_soldier',
    'air_power_ratio',
    'armor_intensity_index',
    'naval_strength_per_coastline',
    'military_burden_index',
    'Power_index_score'
]

df[ratio_cols] = df[ratio_cols].round(6)

int_cols = [
    'active_personnel', 'reserve_personnel', 'total_personnel',
    'total_aircraft', 'fighter_aircraft', 'attack_aircraft',
    'transport_aircraft', 'helicopters', 'tanks',
    'armored_vehicles', 'artillery_units', 'naval_assets',
    'aircraft_carriers', 'population', 'power_index_rank',
    'power_index_rank_gap', 'total_assets', 'year'
]

for col in int_cols:
    if col in df.columns:
        df[col] = df[col].astype(int)

# --- SAVE AS EXCEL FILE (THIS IS THE KEY PART) ---

output_file = "military_kpi_output.xlsx"
df.to_excel(output_file, index=False)

print(f"✅ Excel file created successfully: {output_file}")
print(f"📊 Total rows: {len(df)}")
print("🧮 KPI columns added successfully")
```
