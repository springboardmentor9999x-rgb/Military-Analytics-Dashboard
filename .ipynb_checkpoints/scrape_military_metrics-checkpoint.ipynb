# Module 1: Data Collection and Preparation
**Objective:** Scrape military metrics for 140+ countries from GlobalFirepower.com using a pre-defined list of URLs.
- **Input:** `links_for_military_data.txt` (List of metric pages)
- **Output:** `military_raw_data.csv` (Raw dataset with ~50 indicators)
- **Tools:** Python, Requests, BeautifulSoup, Pandas

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import time
import re
```

```python
def get_data_from_url(url):
    """Visits a page and extracts {Country: Value} pairs."""
    data = []
    try:
        # 1. Fetch page
        page = requests.get(url, headers={'User-Agent': 'Mozilla/5.0'})
        soup = BeautifulSoup(page.content, 'html.parser')

        # 2. Find all country rows (Using the specific class names)
        rows = soup.select("div.recordsetContainer")

        for row in rows:
            country = row.select_one("span.textShadow").text.strip()
            # The value is usually the last 'textLarge' item in the row
            raw_val = row.select("span.textLarge")[-1].text.strip()
            
            # 3. Clean the number (remove commas like '1,200' -> '1200')
            clean_val = re.search(r"-?\d+\.?\d*", raw_val.replace(",", ""))
            if clean_val:
                data.append({'Country': country, 'Value': float(clean_val.group())})
                
    except Exception as e:
        print(f"  Error: {e}")
    
    return pd.DataFrame(data)
```

```python
# --- MAIN PROGRAM ---

print("Starting Scraper...")
final_df = pd.DataFrame()

# 1. Read the text file (Filtering only lines with arrows)
with open('links_for_military_data.txt', 'r', encoding='utf-8') as f:
    lines = [line.strip() for line in f if '→' in line]

# 2. Loop through each link
for line in lines:
    # Split "URL → Column_Name"
    url, col_name = line.split('→')
    url = url.strip().replace('- ', '') # Clean URL
    col_name = col_name.strip()          # Clean Column Name
    
    print(f"Scraping: {col_name}")
    
    # Fetch data
    temp_df = get_data_from_url(url)
    
    if not temp_df.empty:
        # Rename generic 'Value' to the specific column name (e.g., 'total_tanks')
        temp_df.rename(columns={'Value': col_name}, inplace=True)
        
        # Merge into the master list
        if final_df.empty:
            final_df = temp_df
        else:
            final_df = pd.merge(final_df, temp_df, on='Country', how='outer')
            
    time.sleep(1) # Pause specifically to avoid getting blocked
```

```python
# 3. Save to CSV
final_df.sort_values('Country').to_csv('military_raw_data.csv', index=False)
print("Success! Data saved to military_raw_data.csv")
```

# Module 2: Data Cleaning and Structuring
**Objective:** Clean the raw scraped data, handle missing values, and standardize formats for analysis.
- **Input:** `military_raw_data.csv`
- **Output:** `military_cleaned.csv`

```python
import os

# Configuration
INPUT_FILE = 'military_raw_data.csv'
OUTPUT_FILE = 'military_cleaned.csv'

# Check if file exists
if not os.path.exists(INPUT_FILE):
    print(f"Error: {INPUT_FILE} not found. Please run the scraper first.")
else:
    print(f"Ready to clean: {INPUT_FILE}")
```

```python
# Load the dataset
df = pd.read_csv(INPUT_FILE)

# Display initial info
print(f"Initial Shape: {df.shape}")
print("\n--- Missing Values (Top 10 Columns) ---")
print(df.isnull().sum().sort_values(ascending=False).head(10))

df.head()
```

```python
# 1. Fill Missing Values
# Logic: NaN in military data usually implies '0' assets (e.g., no navy for landlocked countries).
print("Filling missing values with 0...")
df.fillna(0, inplace=True)

# 2. Enforce Numeric Formats
# Convert all columns (except 'Country') to numeric types
cols_to_clean = [col for col in df.columns if col != 'Country']

for col in cols_to_clean:
    # Coerce errors to NaN (then 0) to handle any stray text
    df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0)

print("Data cleaning complete.")
```

```python
# Verify there are no more null values
total_missing = df.isnull().sum().sum()

print(f"Total Missing Values: {total_missing}")
if total_missing == 0:
    print("✅ Success: Dataset is clean.")
else:
    print(f"⚠️ Warning: There are still {total_missing} missing values.")

# Check specific column types (optional)
df.info()
```

```python
# Save to CSV
df.to_csv(OUTPUT_FILE, index=False)

print(f"Cleaned dataset saved to: {OUTPUT_FILE}")
```
