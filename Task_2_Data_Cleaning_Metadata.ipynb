```python
import pandas as pd
import numpy as np
```

```python
input_path = "output/global_military_raw_data_with_metadata.csv"

df = pd.read_csv(input_path)

df.head()
```

```python
df.info()
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)
```

```python
df.replace([np.inf, -np.inf], 0, inplace=True)
df.fillna(0, inplace=True)
```

```python
output_path = "output/global_military_cleaned_metadata.csv"

df.to_csv(output_path, index=False)

print("✅ Task 2 (Metadata Cleaning) completed successfully!")
```
