#Module 2: Data Cleaning and Structuring  



```python
import pandas as pd

df = pd.read_csv('/content/military_raw_data.csv')
print("DataFrame reloaded successfully. First 5 rows:")
df.head()
```

## Standardize Column Names

### Subtask:
Convert all column names to a consistent, standardized format (snake_case) for improved readability and usability.


**Reasoning**:
To standardize the column names to snake_case, I will apply a series of string operations: first converting to lowercase, then replacing spaces and special characters with underscores, and finally cleaning up leading/trailing and multiple underscores. After these transformations, I will print the new column names to verify the changes.



```python
import re

# Convert all column names to lowercase
df.columns = df.columns.str.lower()

# Replace spaces and special characters with underscores
df.columns = df.columns.str.replace(r'[^a-z0-9]+', '_', regex=True)

# Remove leading/trailing underscores and replace multiple underscores with a single one
df.columns = df.columns.str.strip('_').str.replace(r'_{2,}', '_', regex=True)

print("Column names standardized to snake_case:")
print(df.columns.tolist())
```

## Clean and Convert Numeric Metrics

Iterate through columns that represent metrics, remove non-numeric characters, and convert these columns to a suitable numeric data type.

**Reasoning**:
To identify and convert numeric metric columns, I will first define the columns to exclude (categorical/identifier). Then, I will iterate through the remaining columns, cleaning them by removing non-numeric characters and converting them to a numeric data type, coercing errors. This ensures that all numerical data is in a clean and consistent format, ready for analysis and imputation.



```python
import numpy as np

# Define columns that should NOT be treated as numeric metrics
# These are categorical or identifier columns
exclude_columns = ['country', 'capital_city', 'region', 'continent', 'alliance', 'year']

# Iterate through all columns and process them
for col in df.columns:
    if col not in exclude_columns:
        # Check if the column is of object type (potential string values that need cleaning)
        if df[col].dtype == 'object':
            # Remove non-numeric characters (like '$', ',', '%')
            df[col] = df[col].astype(str).str.replace(r'[$,%]', '', regex=True)
            # Handle potential 'M' (Millions) or 'B' (Billions) suffixes by multiplying
            def convert_large_numbers(value):
                if isinstance(value, str):
                    value = value.strip()
                    if value.endswith('B'):
                        return float(value[:-1]) * 1_000_000_000
                    elif value.endswith('M'):
                        return float(value[:-1]) * 1_000_000
                return value
            df[col] = df[col].apply(convert_large_numbers)

        # Convert the column to numeric, coercing any errors to NaN
        # This will convert any remaining strings that couldn't be cleaned, or non-numeric types
        # and also ensure all numeric columns are of a type that can hold NaNs (float)
        df[col] = pd.to_numeric(df[col], errors='coerce')

print("DataFrame Info after cleaning and converting numeric metrics:")
df.info()

print("\nFirst 5 rows of DataFrame after cleaning and converting numeric metrics:")
df.head()
```

## Handle Missing Values

### Subtask:
Impute missing values in the `capital_cities_by_total_population` and `navy_force_by_tonnage` columns using their respective medians.

**Reasoning**:
To impute missing values in `capital_cities_by_total_population` and `navy_force_by_tonnage` columns, I will calculate the median for each column, then use the `fillna()` method to replace NaN values with their respective medians. Finally, I will display the number of missing values for these columns to confirm the imputation.



```python
columns_to_impute = ['capital_cities_by_total_population', 'navy_force_by_tonnage']

for col in columns_to_impute:
    if df[col].isnull().any(): # Check if there are any missing values
        median_value = df[col].median()
        df[col] = df[col].fillna(median_value)
        print(f"Missing values in '{col}' imputed with median: {median_value}")
    else:
        print(f"No missing values found in '{col}'.")

print("\nMissing values after imputation for specified columns:")
print(df[columns_to_impute].isnull().sum())

print("\nDataFrame Info after imputation:")
df.info()
```
