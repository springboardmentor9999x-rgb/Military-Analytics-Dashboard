```python
# Task 2: Data Cleaning & Structuring
## Unified Military Analytics & Comparison Dashboard

###This notebook cleans and standardizes raw military data scraped from GlobalFirepower.com.
```

```python
import pandas as pd
import numpy as np
```

```python
# Load raw scraped data
raw_path = "output/global_military_raw_data.csv"
df = pd.read_csv(raw_path)

# Basic inspection
df.head()
```

```python
df.info()
```

```python
# Clean column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)
df.columns
```

```python
# Convert all metric columns to numeric (except country)
for col in df.columns:
    if col not in ["country"]:
        df[col] = pd.to_numeric(df[col], errors="coerce")

df.info()
```

```python
# Check missing values
missing_percent = df.isnull().mean() * 100
missing_percent.sort_values(ascending=False).head(10)
```

```python
df.fillna(0, inplace=True)

# Verify
df.isnull().sum().sum()
```

```python
# Check missing values
missing_percent = df.isnull().mean() * 100
missing_percent.sort_values(ascending=False).head(10)
```

```python
# Number of countries
print("Total countries:", df.shape[0])

# Number of features
print("Total features:", df.shape[1])

df.describe().T.head()
```

```python
# Save cleaned dataset
clean_path = "output/military_cleaned.csv"
df.to_csv(clean_path, index=False)

print("✅ Cleaned dataset saved successfully!")
```
